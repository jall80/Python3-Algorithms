```python
import pandas as pd
```

```python
DF1 = pd.DataFrame({'AUTOS': ['Nissan', 'Ford', 'Audi'],
'COLOR': ['blanco', 'Azul', 'Rojo']})
```

```python
DF2 = pd.DataFrame({'AUTOS': ['Toyota', 'Ford', 'Audi'],
'MODELO': ['2018', '2020', '2022']})
```

```python
pd.merge(DF1, DF2, on='AUTOS', how='outer')
```

```python
pd.merge(DF1, DF2, on='AUTOS', how='left')
```

```python
pd.merge(DF1, DF2, on='AUTOS', how='right')
```
