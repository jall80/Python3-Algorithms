```python

print('Find the youngest employee in each department')


### This method has a problem, it only will take the first min Age, if two or more people have same min age, just one will be taken

ids = df.groupby('Department')['Age'].idxmin().values

df.iloc[ids][['Department', 'Name', 'Age']]

```

```python
print('Find the youngest employee in each department - second')

### This method has a problem, it only will take the first min Age, if two or more people have same min age, just one will be taken

df.groupby('Department')['Age'].nsmallest(1)


```

```python
print('Find the youngest employee in each department - third')

min_age_per_dept = df.groupby('Department')['Age'].transform('min')

print(min_age_per_dept)

df[df['Age'] == min_age_per_dept][['Department', 'Name', 'Age']].sort_values('Department')

```

```python
print('Show the employees who earn more than their department’s average salary')

dep_mean = df.groupby('Department')['Annual_Salary'].mean()

print(dep_mean)


df_merged = df.merge(dep_mean, on = 'Department')


df_merged = df_merged.rename(columns={'Annual_Salary_x': 'Annual_Salary', 'Annual_Salary_y': 'Dep_Avg_Salary'})

df_merged[df_merged['Annual_Salary'] > df_merged['Dep_Avg_Salary']][['Name', 'Department', 'Annual_Salary', 'Dep_Avg_Salary']]
```

```python
print('Show the employees who earn more than their department’s average salary - BETTER')


df['Dep_Avg_Salary'] = df.groupby('Department')['Annual_Salary'].transform('mean')
df[df['Annual_Salary'] > df['Dep_Avg_Salary']][['Name', 'Department', 'Annual_Salary', 'Dep_Avg_Salary']]

```
