```python
print('Inner join')

df1 = pd.DataFrame({
    "id": [1, 2, 3],
    "nombre": ["Ana", "Luis", "Marta"]
})

df2 = pd.DataFrame({
    "id": [1, 2, 4],
    "nota": [90, 85, 75]
})

df_inner = pd.merge(df1, df2, on='id', how='inner')

df_inner

```

```python
print('Left join')

df_left = pd.merge(df1, df2, on='id', how='left')
df_left
```

```python
print('Right join')

df_right = pd.merge(df1, df2, on='id', how='right')
df_right
```

```python
print('Outer join')

df_outer = pd.merge(df1, df2, on='id', how='outer')
df_outer 
```

```python
print('Merge different columns names')

df3 = pd.DataFrame({
    "codigo": [1, 2, 3],
    "ciudad": ["San José", "Cartago", "Heredia"]
})

df_merge = pd.merge(df1, df3, left_on="id", right_on="codigo", how="inner")
print(df_merge)
```

```python
print('Concat')

df_concat = pd.concat([df1, df2], axis=1)
df_concat
```
