```python
## Adding Coluns in data frame

df_pyspark = df_pyspark.withColumn('Experience After two years', df_pyspark['Experience']+2)
df_pyspark.show()
```

```python
### DROP the columns

df_pyspark.drop('Experience After two years')
df_pyspark.show()
```

```python
### Rename the columns

df_pyspark.withColumnRenamed('Name', 'New_name').show()
```
